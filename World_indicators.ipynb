```python
import requests, pandas as pd

PAYS = "ZAF;KEN;NGA;EGY;MAR;CIV;RWA;SEN;GHA;CPV"

def wb(indicator, pays=PAYS, debut=2000, fin=2025):
    url = f"https://api.worldbank.org/v2/country/{pays}/indicator/{indicator}"
    r = requests.get(url, params={"format": "json", "date": f"{debut}:{fin}",
                                  "per_page": 1000}, timeout=30)
    r.raise_for_status()
    _, rows = r.json()
    return pd.DataFrame(
        [{"iso3": d["countryiso3code"], "pays": d["country"]["value"],
          "annee": int(d["date"]), indicator: d["value"]} for d in rows])

croissance_hab = wb("NY.GDP.PCAP.KD.ZG")   # croissance du PIB/hab. (%)
internet       = wb("IT.NET.USER.ZS")      # utilisateurs d'Internet (% pop.)
df = croissance_hab.merge(internet, on=["iso3", "pays", "annee"])

# FMI : croissance du PIB réel, historique + projections
fmi = requests.get(
    "https://www.imf.org/external/datamapper/api/v1/NGDP_RPCH/ZAF/KEN/NGA",
    timeout=30).json()["values"]["NGDP_RPCH"]
```
